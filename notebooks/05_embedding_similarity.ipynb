For each pretrained backbone (no trained head needed), traces the mean cosine similarity between perturbed test embeddings and clean train embeddings across each perturbation's range, with joint t-SNE plots (dissertation section 4.1).

**Writes:** `Stats/<dataset>/<model>/embeddings/<perturbation>_stats.xlsx`, `<perturbation>_tsne.xlsx` and plots, and `Stats/<dataset>/similarity_aggregate/<dataset>_similarity_grid.png`.
