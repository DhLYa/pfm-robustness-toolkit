# 1 · Train classification heads

Trains a binary head on each model for one dataset (backbone frozen; ResNets fine-tuned end to end), with early stopping on validation ROC-AUC.

**Writes:** `Models/<dataset>/<weight_key>_<dataset>.pth` (head `state_dict`; whole model for ResNets).

Paths are resolved automatically (repo root = folder with `pyproject.toml`). If `Data/`, `Models/` and `Stats/` live elsewhere (e.g. on the cluster), set `PFM_STORAGE=/path/to/storage` before starting Jupyter. Gated Hugging Face models need `HF_TOKEN` set.

```python
from pfm_robustness import DATASETS, MODELS
from pfm_robustness.train import train_model

print("Datasets:", list(DATASETS))
print("Models:  ", list(MODELS))
```

## Configure

```python
DATASET = "NCT"
MODEL_NAMES = list(MODELS)   # or e.g. ["ResNet18", "UNI2"]
EPOCHS = 10                  # early stopping (patience 2) usually ends sooner
```

## Train

```python
for name in MODEL_NAMES:
    train_model(DATASET, name, epochs=EPOCHS, show_class_counts=True)
```
