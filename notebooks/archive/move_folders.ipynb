```python
# ==== Move augmentation folders into old_images/ (for all datasets & models) ====
import os
import shutil

# --- configure ---
PATH = "../.."  # <-- set this
DATASETS = ["NCT", "PANDA", "PanNuke", "PatchCamelyon"]  # as per your screenshot
FOLDER_LIST = ["blur", "brightness", "crop", "hed", "jpeg", "pixel",
               "random_stain", "rotate", "stain", "zoomin", "zoomout"]
DRY_RUN = False  # set to False to perform the moves

stats_root = os.path.join(PATH, "Stats")
assert os.path.isdir(stats_root), f"Stats not found at {stats_root}"

moved, skipped_missing, skipped_exists, skipped_notdir = 0, 0, 0, 0
log = []

for ds in DATASETS:
    ds_path = os.path.join(stats_root, ds)
    if not os.path.isdir(ds_path):
        log.append(f"[dataset-missing] {ds_path}")
        continue

    # iterate all model folders present under this dataset
    for model_name in sorted(os.listdir(ds_path)):
        model_path = os.path.join(ds_path, model_name)
        if not os.path.isdir(model_path):
            skipped_notdir += 1
            continue

        dest_root = os.path.join(model_path, "old_images")
        if not DRY_RUN:
            os.makedirs(dest_root, exist_ok=True)

        for folder in FOLDER_LIST:
            src = os.path.join(model_path, folder)
            if not os.path.isdir(src):
                skipped_missing += 1
                log.append(f"[missing] {ds}/{model_name}/{folder}")
                continue

            dst = os.path.join(dest_root, folder)
            if os.path.exists(dst):
                skipped_exists += 1
                log.append(f"[exists]  {ds}/{model_name}/old_images/{folder} (skipped)")
                continue

            log.append(f"[move]    {ds}/{model_name}/{folder}  ->  {ds}/{model_name}/old_images/{folder}")
            if not DRY_RUN:
                shutil.move(src, dst)
                moved += 1

mode = "DRY RUN (no changes made)" if DRY_RUN else "EXECUTED (folders moved)"
print(f"\n=== Move report — {mode} ===")
print(f"Moved: {moved} | Skipped missing: {skipped_missing} | Skipped existing dest: {skipped_exists} | Skipped not-dir: {skipped_notdir}")
print("\nExamples:")
for line in log[:25]:
    print(line)
if len(log) > 25:
    print(f"... ({len(log) - 25} more)")
```

```python
# JUPYTER CELL — Delete "old_images" under PATH/Stats/{dataset}/{model}
# ---------------------------------------------------------------------------------
# Expects a variable named PATH pointing to the root directory that contains "Stats".
# If PATH is not already defined above, uncomment and set it here:
# PATH = "/absolute/or/relative/root"

import os
import shutil
import sys

PATH = "../.."

# Safety toggle: start with a dry run (no deletions). Set to False to actually delete.
DRY_RUN = False

def delete_old_images_at_path(root_path: str, dry_run: bool = True):
    base_dir = os.path.join(root_path, "Stats")
    if not os.path.isdir(base_dir):
        print(f"[ERROR] 'Stats' directory not found at: {base_dir}")
        return

    scanned_models = deleted = missing = errors = 0

    for dataset_name in sorted(os.listdir(base_dir)):
        ds_path = os.path.join(base_dir, dataset_name)
        if not os.path.isdir(ds_path):
            continue

        for model_name in sorted(os.listdir(ds_path)):
            model_path = os.path.join(ds_path, model_name)
            if not os.path.isdir(model_path):
                continue

            scanned_models += 1
            target = os.path.join(model_path, "old_images")

            if os.path.isdir(target):
                action = "Would delete" if dry_run else "Deleting"
                print(f"{action}: {target}")
                if not dry_run:
                    try:
                        shutil.rmtree(target)
                        deleted += 1
                    except Exception as e:
                        errors += 1
                        print(f"[ERROR] Failed to delete {target}: {e}")
            else:
                missing += 1

    print(
        f"\n[SUMMARY] Model folders scanned: {scanned_models} | "
        f"deleted: {deleted} | no 'old_images': {missing} | errors: {errors}"
    )

# ---- Run (uncomment the next line to execute) ----
delete_old_images_at_path(PATH, dry_run=DRY_RUN)

# Tip: first run with DRY_RUN=True to preview actions, then set DRY_RUN=False and rerun.
```
