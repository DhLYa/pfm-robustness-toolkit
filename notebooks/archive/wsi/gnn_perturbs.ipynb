```python
def extract_and_convert_wsi(wsi_path, target_mag, out_dir):
    slide = openslide.OpenSlide(wsi_path)
    base_mag = float(slide.properties.get('openslide.objective-power', 20))
    mpp_base = float(slide.properties.get('openslide.mpp-x', 0.5))
    mpp_target = mpp_base * (base_mag / target_mag)

    level = slide.get_best_level_for_downsample(base_mag / target_mag)
    W, H = slide.level_dimensions[level]
    region = slide.read_region((0, 0), level, (W, H)).convert("RGB")

    os.makedirs(out_dir, exist_ok=True)
    base = os.path.splitext(os.path.basename(wsi_path))[0]
    png_full = os.path.join(out_dir, f"{base}_extracted_{target_mag}x.png")
    region.save(png_full)

    converter = AnyToTiffConverter(job_dir=out_dir, bigtiff=True)
    converter.process_file(png_full, mpp=mpp_target, zoom=1)
    return region, mpp_target

def run_wsi_perturbation_pipeline(
    wsi_dir, target_mag=2.5,
    out_root="wsi_pipeline",
    device="cuda:0"
):
    print("WSI dir:", wsi_dir, "Exists?", os.path.exists(wsi_dir))
    if os.path.exists(wsi_dir):
        print("Contents:", os.listdir(wsi_dir))
    else:
        raise FileNotFoundError(wsi_dir)

    os.makedirs(out_root, exist_ok=True)
    converter = AnyToTiffConverter(job_dir=out_root, bigtiff=True)
    model = model_evals[0]["load_func"]().to(device).eval()
    to_tensor = transforms.ToTensor()

    tif_paths = glob.glob(os.path.join(wsi_dir, "*.tif"))
    print(f"Found {len(tif_paths)} .tif files")

    for wsi_path in tif_paths:
        base = os.path.splitext(os.path.basename(wsi_path))[0]
        print("\nProcessing:", base)

        region, mpp_target = extract_and_convert_wsi(
            wsi_path, target_mag, out_root
        )

        x = to_tensor(region)  # [C, H, W], keep on CPU for now
        C_, H, W = x.shape
        h2, w2 = H//2, W//2
        quarters = [
            x[:, :h2, :w2],
            x[:, :h2, w2:],
            x[:, h2:, :w2],
            x[:, h2:, w2:]
        ]

        for pert_name, specs in perturbations.items():
            print(" →", pert_name)
            attack = specs["TransformOptimiser"](
                model, specs["Transform"],
                specs["optimiser_params"], specs["transform_params"],
                device=device
            )

            start, end = sweep_ranges[pert_name]
            step = step_sizes[pert_name]
            values = np.arange(start, end + step, step)

            for v in values:
                attack.hyperparameters[sweep_params[pert_name]] = v

                adv_q_list = []
                for q in quarters:
                    q_gpu = q.unsqueeze(0).to(device)   # shape [1, C, h2, w2]
                    label = torch.tensor([0], dtype=torch.long, device=device)
                    _, adv_single = attack.optimise(q_gpu, targets=label, reset_weights=True)
                    adv_q_list.append(adv_single[0].cpu())

                adv_full = torch.zeros_like(x)
                adv_full[:, :h2, :w2] = adv_q_list[0]
                adv_full[:, :h2, w2:] = adv_q_list[1]
                adv_full[:, h2:, :w2] = adv_q_list[2]
                adv_full[:, h2:, w2:] = adv_q_list[3]

                out_dir = os.path.join(out_root, "all_perturbations", pert_name, base)
                os.makedirs(out_dir, exist_ok=True)
                png_out = os.path.join(out_dir, f"{base}_{v:.3f}.png")
                # img_arr = (adv_full.permute(1, 2, 0).numpy() * 255).astype(np.uint8)
                img_arr = (adv_full.detach().permute(1, 2, 0).cpu().numpy() * 255).astype(np.uint8)
                Image.fromarray(img_arr).save(png_out)

                converter.process_file(png_out, mpp=mpp_target, zoom=1)

    print("\n✅ All WSIs processed with SERIAL quartered perturbations.")

# 3) Run it
run_wsi_perturbation_pipeline(
    wsi_dir=os.path.join(PATH, "data", "CAMELYON", "test"),
    target_mag=2.5,
    out_root=os.path.join(PATH, "data", "CAMELYON", "test", "wsi_pipeline"),
    device="cuda:0"
)
```
