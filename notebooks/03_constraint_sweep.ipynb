# 3 · Constraint sweeps and PPI

Sweeps each perturbation in `SWEEPS` from no perturbation to its maximum (dissertation Table 1), recording adversarial ROC-AUC per step, then computes the Perturbation Performance Index (area under the ROC-AUC vs normalised-constraint curve). Needs the trained heads from notebook 1.

**Writes:**
- `Stats/<dataset>/<model>/constraint_sweep/<model>_<perturbation>_sweep.xlsx` (and a plot)
- `Stats/<dataset>/constraint_sweep/<perturbation>_aggregate.png`
- `Stats/<dataset>/constraint_sweep/ppi_aggregate.xlsx`, `ppi_mean_bar.png`, `ppi/` (heatmap, PPI vs parameters)

Paths are resolved automatically (repo root = folder with `pyproject.toml`). If `Data/`, `Models/` and `Stats/` live elsewhere (e.g. on the cluster), set `PFM_STORAGE=/path/to/storage` before starting Jupyter. Gated Hugging Face models need `HF_TOKEN` set.

```python
from pfm_robustness import MODELS, SWEEPS
from pfm_robustness.ppi import calculate_ppi_aggregate
from pfm_robustness.sweep import aggregate_constraint_sweeps, run_sweeps

for name, sw in SWEEPS.items():
    print(f"{name:13s} {sw.param:8s} {sw.range}  steps={sw.steps}")
```

## Configure

The test subset size comes from `DATASETS[DATASET].budgets["sweep"]`; pass `test_multiplier=` to `run_sweeps` to override it.

```python
DATASET = "NCT"
MODEL_NAMES = list(MODELS)
RUNS = 3
```

## Run the sweeps

```python
run_sweeps(DATASET, MODEL_NAMES, runs=RUNS)
```

## Aggregate plots and PPI

Reads whatever sweep workbooks exist for this dataset, so it can be re-run on its own.

```python
aggregate_constraint_sweeps(DATASET, MODEL_NAMES)
ppi_mean, ppi_std = calculate_ppi_aggregate(DATASET, MODEL_NAMES)
ppi_mean.round(3)
```
